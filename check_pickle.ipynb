# 🔍 Check Pickle Data

Gunakan notebook ini untuk memeriksa isi file hasil transformasi.

### 1. Lihat Daftar File
Jalankan cell di bawah ini untuk melihat daftar file yang tersedia.

```python
from glob import glob
import os

# Scan semua file pickle
pickle_files = glob('**/*.pkl', recursive=True)
print("Daftar file yang bisa kamu periksa:")
print("-" * 40)
for i, f in enumerate(pickle_files):
    print(f"[{i}] {f}")
print("-" * 40)
print("Silakan pilih nomor index untuk digunakan di Cell 2.")
```

### 2. Tampilkan Isi File
Ganti nilai `pilih_index` dengan nomor yang kamu inginkan, lalu jalankan cell ini.

```python
import pickle
import pandas as pd

pilih_index = 3  # <--- GANTI ANGKA INI SESUAI PILIHANMU

if 'pickle_files' in locals() and 0 <= pilih_index < len(pickle_files):
    target_path = pickle_files[pilih_index]
    print(f"📂 Membuka file: {target_path}\n")
    
    try:
        with open(target_path, "rb") as f:
            data = pickle.load(f)
        
        if isinstance(data, dict):
            print(f"✅ Berhasil! Ditemukan {len(data)} tabel.\n")
            for table_name, df in data.items():
                print(f"📌 Tabel: {table_name} ({len(df)} baris)")
                if hasattr(df, 'head') and not df.empty:
                    display(df.head(2))
                elif df.empty:
                    print("   ⚠️ Tabel ini kosong")
        else:
            print("Data bukan format dictionary, berikut isinya:")
            print(data)
            
    except NotImplementedError:
        print("❌ ERROR: File ini masih pakai tipe data 'StringDtype' yang bermasalah di Python 3.13.")
        print("💡 SOLUSI: Jalankan ulang script transformasinya dulu agar filenya diperbarui ke format aman.")
    except Exception as e:
        print(f"❌ Gagal membuka file: {e}")
else:
    print("❌ Index tidak valid atau Cell 1 belum dijalankan!")
```
